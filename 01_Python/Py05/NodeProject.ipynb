## 파트 1: 사각형 넓이 구하기 클래스 (Square)

```python
class Square:
    def __init__(self):
        # 사각형 유형 입력 및 안내 문구 분기
        self.square = int(input('넓이를 구하고 싶은 사각형의 숫자를 써주세요.\n 1.직사각형 2.평행사변형 3.사다리꼴 \n >>>'))

        if self.square == 1:
            print('직사각형 함수는 rect()입니다.')
        elif self.square == 2:
            print('평행사변형 함수는 par()입니다.')
        elif self.square == 3:
            print('사다리꼴 함수는 trape()입니다.')
        else:
            print('1, 2, 3 중에서 다시 입력해주세요')

    # 직사각형 넓이 계산 및 출력
    def rect(self):
        width, vertical = map(int, input('가로, 세로 사이즈(숫자)를 입력하세요. 예시 : 5,10\n >>>').split(','))
        area = width * vertical
        return '직사각형의 넓이는 : ' + str(area)

    # 평행사변형 넓이 계산 및 출력
    def par(self):
        base, height = map(int, input('밑변, 높이 사이즈(숫자)를 입력하세요. 예시 : 5,8\n >>>').split(','))
        area = base * height
        return '평행사변형의 넓이는 : ' + str(area)

    # 사다리꼴 넓이 계산 및 출력 (우선순위 괄호 반영)
    def trape(self):
        base, top, height = map(int, input('밑변, 윗변, 높이 사이즈(숫자)를 입력하세요. 예시 : 4,5,6\n >>>').split(','))
        area = (base + top) * height / 2
        return '사다리꼴의 넓이는 : ' + str(area)
        
# ==================== 실전 구동 영역 ====================
# 사각형 계산기 객체 생성
calculator = Square()

# 입력 번호에 따른 메서드 분기 실행
if calculator.square == 1:
    print(calculator.rect())
elif calculator.square == 2:
    print(calculator.par())
elif calculator.square == 3:
    print(calculator.trape())
```

## 파트 2: 키오스크 시스템 클래스 (Kiosk)

```python
class Kiosk:
    def __init__(self):
        # 메뉴 및 가격 인스턴스 속성 등록
        self.menu = ['americano', 'latte', 'mocha', 'yuza_tea', 'green_tea', 'choco_latte']
        self.price = [2000, 3000, 3000, 2500, 2500, 3000]
        
        # 최종 주문서 출력을 위한 장바구니 리스트 초기화
        self.order_menu = []
        self.order_price = []

    # 메뉴 출력 메서드
    def menu_print(self):
        print('\n⟝' + '=' * 34 + '⟞')
        print('|' + ' ' * 7 + '로그픽 카페 메뉴판' + ' ' * 9 + '|')
        print('⟝' + '-' * 34 + '⟞')
        
        for i in range(len(self.menu)):
            menu_name = self.menu[i].ljust(12) 
            print(f"|  [{i + 1}] {menu_name}  :  {self.price[i]}원  |")
            
        print('⟝' + '=' * 34 + '⟞\n')

    # 주문 메서드
    def menu_select(self):
        n = 0
        while n < 1 or len(self.menu) < n:
            n = int(input("안내: 음료 번호를 입력하세요 : "))

            if 1 <= n and n <= len(self.menu):
                self.price_sum = self.price[n-1]
            else:  
                print("오류: 없는 메뉴입니다. 다시 주문해 주세요.\n")
                
        # 온도 입력 전 깔끔하게 한 줄 띄우기
        print()  
        t = 0  
        while t != 1 and t != 2:  
            t = int(input("안내: HOT 음료는 1을, ICE 음료는 2를 입력하세요 : "))
            if t == 1:
                self.temp = "HOT"
            elif t == 2:
                self.temp = "ICE"
            else:    
                print("오류: 1과 2 중 하나를 입력하세요.\n")
                
        # 주문 중간 확인 전후 여백 배치
        print(f"\n[알림] {self.temp} {self.menu[n-1]} ({self.price_sum}원) 상품이 장바구니에 담겼습니다.\n")
        
        # 장바구니 리스트에 최종 주문 데이터 누적
        self.order_menu.append(f"{self.temp} {self.menu[n-1]}")
        self.order_price.append(self.price_sum)

    # 지불 메서드
    def pay(self):
        payment_method = 0
        while payment_method not in ['cash', 1, 'card', 2]:
            payment_method = input("안내: 결제 수단을 선택하세요 (현금: cash 또는 1 / 카드: card 또는 2) : ")
            
            if payment_method.isdigit():
                payment_method = int(payment_method)

            print()
            if payment_method == 'cash' or payment_method == 1:
                print('알림: 직원을 호출하겠습니다. 잠시만 기다려주세요.')
            elif payment_method == 'card' or payment_method == 2:
                print('알림: IC칩 방향에 맞게 카드를 꽂아주세요.')
            else:
                print('오류: 잘못된 입력입니다. 다시 결제를 시도해 주세요.\n')
            print()

    # 주문서 출력 메서드
    def table(self):
        print('⟝' + '=' * 34 + '⟞')
        print('|' + ' ' * 9 + '주 문 영 수 증' + ' ' * 11 + '|')
        print('⟝' + '-' * 34 + '⟞')
        
        for i in range(len(self.order_menu)):
            order_item = self.order_menu[i].ljust(18)
            print(f"|  {order_item}  :  {self.order_price[i]}원  |")

        print('⟝' + '-' * 34 + '⟞')
        total_price = f"{sum(self.order_price)}원"
        print(f"|  * 총 합계 금액 * :  {total_price.rjust(7)}  |")
        print('⟝' + '=' * 34 + '⟞\n')


# ==================== 실전 구동 영역 ====================
# 키오스크 객체 생성
cafe_kiosk = Kiosk()

# 초기 메뉴판 출력
cafe_kiosk.menu_print()

# 추가 주문 제어를 위한 무한 루프
while True:
    # 음료 선택 및 장바구니 담기 진행
    cafe_kiosk.menu_select()
    
    # 사용자 추가 주문 여부 확인
    repeat = input("확인: 추가 주문을 하시겠습니까? (y: 추가 주문 / n: 주문 완료) : ").lower()
    
    # 'n' 입력 시 루프 탈출 및 결제 단계로 전환
    if repeat == 'n':
        print("\n----------------------------------")
        print("시스템: 주문이 마감되었습니다. 결제 단계로 이동합니다.")
        print("----------------------------------\n")
        break
    print()  # 추가 주문 시 가독성을 위한 줄 바꿈

# 결제 수단 선택 및 처리
cafe_kiosk.pay()

# 최종 주문 영수증 출력
cafe_kiosk.table()
```
