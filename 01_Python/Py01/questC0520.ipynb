# Q1-1 회문 판독기

```python
# 회문 판독기 함수
def is_palindrome(word):
    return word == word[::-1]

# 변수 선언
target1 = "기러기"
target2 = "hello"

# 결과 출력
print(is_palindrome(target1))
print(is_palindrome(target2))
```

# Q1-2 회문 분류기

```python
# 회문 분류기 함수
def palindrome_classifier(word_list):

    # 분류용 딕셔너리 초기화
    result_dict = {
        '회문O': [], 
        '회문X': []
    }

    # 반복문으로 리스트에서 하나씩 검사
    for word in word_list:
        if is_palindrome(word) == True:
            result_dict['회문O'].append(word) # 회문이면 O
        else:
            result_dict['회문X'].append(word) # 회문이 아니면 X

    # 갯수 카운트
    total_count = len(word_list) # 전체 갯수 카운트
    palindrome_count = len(result_dict['회문O']) # 회문 갯수 카운트

    # 회문 비율 계산
    if total_count > 0:
        palindrome_ratio = palindrome_count / total_count
    else:
        palindrome_ratio = 0.0

    # 결과값 반환
    return (result_dict, palindrome_count, palindrome_ratio)

# 리스트 변수 선언
targets = ['기러기', '까마귀', '토마토', '스윗소스', '스위스', '인도인', '한국인', '별똥별', '우영우', '우향우']

# 결과 출력
print(palindrome_classifier(targets))
```

# Q2-1 성적 데이터화

```python
# 딕셔너리 변수 선언
scores = {
    '유원영': {
        '국어': 94, 
        '영어': 91, 
        '수학': 89
    },
    '설지민': {
        '국어': 88, 
        '영어': 96, 
        '수학': 90
    },
    '신윤아': {
        '국어': 86, 
        '영어': 88, 
        '수학': 77
    },
    '장윤아': {
        '국어': 98, 
        '영어': 76, 
        '수학': 92
    }
}

# 결과 출력
print(scores)
```

# Q2-2 성적 분석기

```python
# 성적 분석하는 함수
def score_analysis(name):

    # 각 과목 평균 점수
    class_kor_avg = 91.50
    class_eng_avg = 87.75
    class_math_avg = 87.00
    class_total_avg = 88.75

    # 입력받은 학생의 과목명
    kor = scores[name]['국어']
    eng = scores[name]['영어']
    math = scores[name]['수학']

    # 전과목 평균 계산
    my_avg = round((kor + eng + math) / 3, 2)

    # 각 과목 평균과의 차이 계산
    kor_diff = kor - class_kor_avg
    eng_diff = eng - class_eng_avg
    math_diff = math - class_math_avg
    avg_diff = my_avg - class_total_avg

    # 결과 출력
    print(f"[{name}] 성적 분석")
    print(f"국어: {kor}점, 학급 평균({class_kor_avg:.2f}점) 대비 {kor_diff:+.2f}")
    print(f"영어: {eng}점, 학급 평균({class_eng_avg:.2f}점) 대비 {eng_diff:+.2f}")
    print(f"수학: {math}점, 학급 평균({class_math_avg}점) 대비 {math_diff:+.2f}")
    print() # 한 줄 띄우기
    print(f"전과목 평균: {my_avg:.2f}점, 학급 평균({class_total_avg:.2f}점) 대비 {avg_diff:+.2f}")

# 함수 입력값
score_analysis("설지민")
```
