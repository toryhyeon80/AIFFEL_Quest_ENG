# Q4. 파이썬을 사용하여 간단한 계산기 프로그램 만들기

```python
# 제곱 연산(**)을 지원 받기 위해 math 모듈 로드
import math

while True:
    
    # 1. 첫 번째 정수 입력받기 (정수가 입력될 때까지 무한 반복)
    while True:
        try:
            num1 = int(input("첫 번째 정수를 입력하세요: "))
            break  # 정수 변환에 성공하면 내부 while문 탈출!
        except ValueError:
            print("잘못된 입력입니다. 정수를 입력해주세요.")
            
    # 2. 연산자 입력받기 (+, -, *, /, **)
    operator = input("연산자를 입력하세요 (+, -, *, /, **): ")

    # 사용자가 지원하지 않는 연산자를 입력한 경우 두 번째 정수를 입력하지 않고 오류 메세지 출력
    if operator not in ['+', '-', '*', '/', '**']:
        print(f"오류: '{operator}'는 지원하지 않는 연산자입니다.")
        print("-" * 30)
        continue  # while문의 맨 처음(첫 번째 정수 입력)으로 즉시 리부트

    # 3. 두 번째 정수 입력받기 (정수가 입력될 때까지 무한 반복)
    while True:
        try:
            num2 = int(input("두 번째 정수를 입력하세요: "))
            break  # 정수 변환에 성공하면 내부 while문 탈출!
        except ValueError:
            print("잘못된 입력입니다. 정수를 입력해주세요.")

    # 4. 연산자별로 조건 분기하여 계산 수행
    try:
        if operator == '+':
            result = num1 + num2
            print(f"결과: {num1} + {num2} = {result}")
            
        elif operator == '-':
            result = num1 - num2
            print(f"결과: {num1} - {num2} = {result}")
            
        elif operator == '*':
            result = num1 * num2
            print(f"결과: {num1} * {num2} = {result}")
            
        elif operator == '/':
            # 나눗셈 연산 시 두 번째 정수가 0이면 ZeroDivisionError 예외
            result = num1 / num2
            print(f"결과: {num1} / {num2} = {result}")
            
        elif operator == '**':
            # math 모듈을 사용하여 제곱 연산 지원 조건 충족!
            result = math.pow(num1, num2)
            print(f"결과: {num1} ** {num2} = {result}")

    except ZeroDivisionError:
        print("오류: 0으로 나눌 수 없습니다.")

    print("-" * 30)
    
    # 5. 계산기를 통해 계속 계산을 할 것인지 입력받기
    while True:
        retry = input("계산을 계속 하시겠습니까? (Y/N): ")
        
        # 사용자가 대문자나 소문자 어떤 걸 입력하든 처리하기 위해 .upper() 활용!
        if retry.upper() == 'N':
            print("계산기 프로그램을 종료합니다.")
            break  # 내부 while문 탈출
        elif retry.upper() == 'Y':
            break  # 다음 계산을 위해 내부 while문만 탈출하고 메인 루프로 복귀!
        else:
            print("잘못된 입력입니다. Y 또는 N만 입력해주세요.")
            
    # 만약 위에서 'N'을 입력했다면 메인 루프 전체를 종료합니다.
    if retry.upper() == 'N':
        break
```
